## 2. Pie chart

Pie charts are used to show percentage or proportional data.
matplotlib provides the pie() method to create pie charts

### It has several advantages:
• Summarizes a large dataset in visual form
• Displays the relative proportions of multiple
classes of data
• Size of the circle is made proportional to the
total quantity


```python
values = [145, 411, 34, 141, 25, 256]

labels = ['China', 'India', 'USA', 'Indonesia', 'Brazil', 'Pakistan']
plt.pie(values,labels= labels)
plt.show();
```

```python
# Data: Country wise Density(pop/km2)

plt.figure(figsize=(10,10))

values = [145, 411, 34, 141, 25, 256]
colors = ['peru','lightgreen','slategrey','lightpink','lightblue','coral']
labels = ['China', 'India', 'USA', 'Indonesia', 'Brazil', 'Pakistan']
explode = (0, 0.1, 0, 0, 0, 0)
plt.pie(values, colors=colors, labels= labels,explode=explode,shadow=True)
plt.title('Population Density per KM2')
plt.legend(labels,loc=0) # 0 for best
plt.show()
```

```python
# Data: Country wise Density(pop/km2)

plt.figure(figsize=(10,10))

values = [145, 411, 34, 141, 25, 256]
colors = ['peru','lightgreen','slategrey','lightpink','lightblue','coral']
labels = ['China', 'India', 'USA', 'Indonesia', 'Brazil', 'Pakistan']
explode = (0, 0.1, 0, 0, 0, 0)
plt.pie(values, colors=colors, labels= labels,explode=explode,counterclock=False,
        shadow=True, autopct='%0.2f%%')
plt.title('Population Density per KM2')
plt.legend(labels,loc=0) # 0 for best
plt.show()
```

## 3. Scatter Plot

A scatter plot is used to graphically display the relationships between variables.
However, to control a plot, it is recommended to use scatter() method.


### It has several advantages:

• Shows the correlation between variables
• Is suitable for large datasets
• Is easy to find clusters
• Is possible to represent each piece of data as a
point on the plot

```python
np.random.seed(123)

# Generate a random float number using a numpy between any float range
wt = np.random.uniform(60.1,68.5, size = 10)
ht = np.random.uniform(124.3,168.7, size = 10)

plt.scatter(wt,ht,c='red',marker='D')
plt.xlabel('weight', fontsize=16)
plt.ylabel('height', fontsize=16)
plt.title('scatter plot - height vs weight',fontsize=20)
plt.show()
```

```python
plt.figure(figsize=(10,5))
np.random.seed(123)

# Generate a random float number using a numpy between any float range
wt_m = np.random.uniform(60.1,68.5, size = 10)
ht_m = np.random.uniform(124.3,168.7, size = 10)

wt_f = np.random.uniform(45.5,61.5, size = 10)
ht_f = np.random.uniform(110.3,155.5, size = 10)

plt.scatter(wt_m,ht_m,c='red',marker='s',label = 'Male' )
plt.scatter(wt_f,ht_f,c='green',marker='*', s = 100 * 2, label = 'Female')
plt.xlabel('weight', fontsize=16)
plt.ylabel('height', fontsize=16)
plt.title('scatter plot - height vs weight',fontsize=20)
plt.legend()
plt.show()
```

## 4. Bar Chart

A bar graph is a pictorial rendition of statistical data in which the independent variable can attain only certain discrete values. The dependent variable may be discrete or continuous. The most common form of bar graph is the vertical bar graph, also called a column graph.

#### The following are advantages of bar graph:
- Show each data category in a frequency distribution
- isplay relative numbers/proportions of multiple categories
- Summarize a large amount of data in a visual, easily intepretable form
- Make trends easier to highlight than tables do
- Estimates can be made quickly and accurately
- Permit visual guidance on accuracy and reasonableness of calculations
- Accessible to a wide audience

```python
City=['Delhi','Beijing','Washington','Tokyo','Moscow']
Price_sqft_in_K=[60,40,70,65,45]

plt.bar(City,Price_sqft_in_K,color='skyblue',edgecolor='yellow')

plt.xlabel('City', fontsize=16)
plt.ylabel('Price_sft_in_K', fontsize=16)
plt.title('Barchart - Price_sft_in_K across cities',fontsize=20)
plt.show()
```

## Horizontal Bar Chart

```python
City=['Delhi','Beijing','Washington','Tokyo','Moscow']

Price_sqft_in_K=[60,40,70,65,45]

plt.barh(City,Price_sqft_in_K,color='skyblue',edgecolor='yellow')

plt.xlabel('City', fontsize=16)
plt.ylabel('Price_sft_in_K', fontsize=16)
plt.title('Barchart - Price_sft_in_K across cities',fontsize=20)
plt.show()
```

## Stacked Bar Chart

A stacked bar chart, also known as a stacked bar graph, is a graph that is used to break down and compare parts of a whole. Each bar in the chart represents a whole, and segments in the bar represent different parts or categories of that whole. Different colors are used to illustrate the different categories in the bar.

```python
City=['Delhi','Beijing','Washington','Tokyo','Moscow']

Price_sqft_in_K_center = [60,40,70,65,45]
Price_sqft_in_K_outer = [45,22,50,52,20]

location = ['center', 'outer']

plt.barh(City,Price_sqft_in_K_center,color='skyblue',edgecolor='yellow')
plt.barh(City,Price_sqft_in_K_outer,color='darkblue',edgecolor='yellow')

plt.xlabel('Price_sft_in_K', fontsize=16)
plt.ylabel('City', fontsize=16)
plt.title('Barchart - Price_sft_in_K across cities',fontsize=20)
plt.legend(location, loc = 0)
plt.show()
```

### Grouped Bar Chart

Grouped bar charts require three columns or rows of data from your DataSet—one for series, one for categories, and one for values. If you have more than two series, each additional series is implemented as an additional bar. For information about value, category, and series data, see Understanding Chart Data.

The following graphic shows you how the data from the series, categories, and values columns in a typical column-based spreadsheet is converted into a vertical grouped bar chart:

```python
city=['Delhi','Beijing','Washington','Tokyo','Moscow']

Price_sqft_in_K_center = [60,40,70,65,15]
Price_sqft_in_K_outer = [45,22,50,52,20]


pos = np.arange(len(city))
print(pos)
w = 0.4

plt.bar(pos,Price_sqft_in_K_center,width = w,color='skyblue',edgecolor='yellow', label = 'City Center')
plt.bar(pos+w,Price_sqft_in_K_outer,width = w,color='darkblue',edgecolor='yellow', label = 'Outskirts')

plt.xticks(pos, city)

plt.xlabel('City', fontsize=16)
plt.ylabel('Price_sft_in_K', fontsize=16)
plt.title('Barchart - Price_sft_in_K across cities',fontsize=20)
plt.legend(loc = 0)
plt.show()
```

## 5. Box plot


Description
A Box and Whisker Plot (or Box Plot) is a convenient way of visually displaying the data distribution through their quartiles.

The lines extending parallel from the boxes are known as the “whiskers”, which are used to indicate variability outside the upper and lower quartiles. Outliers are sometimes plotted as individual dots that are in-line with whiskers. Box Plots can be drawn either vertically or horizontally.

Although Box Plots may seem primitive in comparison to a Histogram or Density Plot, they have the advantage of taking up less space, which is useful when comparing distributions between many groups or datasets.

For some distributions/datasets, you will find that you need more information than the measures of central tendency (median, mean, and mode)

- Boxplots are a standardized way of displaying the distribution of data based on a five number summary (“minimum”, first quartile (Q1), median, third quartile (Q3), and “maximum”).
- median (Q2/50th Percentile): the middle value of the dataset.
- first quartile (Q1/25th Percentile): the middle number between the smallest number (not the “minimum”) and the median of the dataset.
- third quartile (Q3/75th Percentile): the middle value between the median and the highest value (not the “maximum”) of the dataset.
- interquartile range (IQR): 25th to the 75th percentile.
- whiskers (shown in blue)
- outliers (shown as green circles)
    “maximum”: Q3 + 1.5*IQR
    “minimum”: Q1 -1.5*IQR

```python
# Data: Salaries in lacs
fresher = np.round(np.random.uniform(3.5, 6.5, size = 10),1)
lead = np.round(np.random.uniform(5.5, 12.5, size = 10),1)
manager = np.round(np.random.uniform(8.5, 16.5, size = 10),1)
director = np.round(np.random.uniform(18.25, 34, size = 10),1)
vp = np.round(np.random.uniform(27, 50.5, size = 10),1)
cxo = np.round(np.random.uniform(45.5,75, size = 10),1)
```

```python
plt.figure(figsize=(15,10))
box_plot_data=[fresher,lead,manager,director,vp,cxo]
plt.boxplot(box_plot_data)
plt.show()
```

## Add few more properties¶

```python
plt.figure(figsize=(15,10))
box_plot_data=[fresher,lead,manager,director,vp,cxo]
plt.boxplot(box_plot_data,patch_artist=True,
            labels=['fresher','lead','manager','director','vp','cxo'])
plt.show()
```

## Horizontal Box Plot¶

```python
plt.figure(figsize=(15,10))
box_plot_data=[fresher,lead,manager,director,vp,cxo]
plt.boxplot(box_plot_data,patch_artist=True,
            labels=['fresher','lead','manager','director','vp','cxo'],
            vert=0)
plt.show()
```

```python
plt.figure(figsize=(15,10))
box_plot_data=[fresher,lead,manager,director,vp,cxo]

box = plt.boxplot(box_plot_data,patch_artist=True,
                  labels=['fresher','lead','manager','director','vp','cxo'],
                  vert=0)

colors = ['darkblue', 'aqua', 'lime', 'black', 'indigo', 'teal']
for patch, color in zip(box['boxes'], colors):
    patch.set_facecolor(color)

plt.show()
```

## 6. Histogram

Histograms are graphical representations of a probability distribution. A histogram is a
kind of a bar chart.

Using matplotlib and its bar chart function, you can create histogram charts.

#### Advantages of Histogram charts:
    • They display the number of values within a specified interval.
    • They are suitable for large datasets as they can be grouped within the intervals.

```python
values = np.random.randint(24,100, size = 65)
plt.hist(values,bins= 5, histtype='step', align='mid',
         color='g', label='Test Score Data')
plt.legend(loc= 0)
plt.title('Histogram of score')
plt.show()
```

```python
values = np.random.randint(24,100, size = 65)

plt.figure(figsize=(15,5))
plt.hist(values,bins=7, histtype='bar',align='mid', color='c',
         label='Test score Data',edgecolor='black')
plt.legend()
plt.title('Histogram of score')
plt.show()
```

## 2. Line Plot
Line chart is used when you want to emphasize changes in values for one variable (plotted on the vertical axis) for continuous values of a second variable (plotted on the horizontal).

On the horizontal axis, you need a variable that depicts continuous values that have a regular interval of measurement - like minute, hour, day, week, or month etc.,

On the vertical axis, you will report the value of a second numeric variable for points that fall in each of the intervals defined by the horizontal-axis variable.

```python
sns.lineplot(x="size", y="tip",data=tips_df);
```
